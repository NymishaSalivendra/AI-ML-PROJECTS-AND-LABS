## K-Nearest Neighbors (KNN) Classification

```python
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
```

## TASK-4: Train-Test Split

Split the dataset into training (70%) and testing (30%) sets.

```python
# Prepare features and target
X = diabetes_data.drop('Outcome', axis=1)
y = diabetes_data['Outcome']

# Split the data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=101)

print(f"Training set size: {X_train.shape[0]} samples")
print(f"Testing set size: {X_test.shape[0]} samples")
```

## TASK-5: Build and Evaluate KNN Model

Train a KNN classifier with k=1 and evaluate its performance.

```python
# Train KNN with k=1
knn = KNeighborsClassifier(n_neighbors=1)
knn.fit(X_train, y_train)

# Make predictions
y_pred = knn.predict(X_test)

# Evaluate the model
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
print(f"\nAccuracy Score: {accuracy_score(y_test, y_pred)*100:.2f}%")
```

## TASK-6: Finding the Optimal K Value

Test different values of K (from 1 to 40) and plot the error rate to find the optimal K value.

```python
# Store the original unscaled data for later comparison
unchanged_data = X.copy()

# Function to evaluate KNN performance across different K values
def plot_KNN_error_rate(xdata, ydata):
    error_rate = []
    test_scores = []
    train_scores = []

    X_train, X_test, y_train, y_test = train_test_split(xdata, ydata, test_size=0.3, random_state=101)

    for i in range(1, 40):
        knn = KNeighborsClassifier(n_neighbors=i)
        knn.fit(X_train, y_train)
        pred_i = knn.predict(X_test)

        error_rate.append(np.mean(pred_i != y_test))
        train_scores.append(knn.score(X_train, y_train))
        test_scores.append(knn.score(X_test, y_test))

    # Plot error rate
    plt.figure(figsize=(12, 8))
    plt.plot(range(1, 40), error_rate, color='blue', linestyle='dashed', marker='o',
             markerfacecolor='red', markersize=8)
    plt.title('Error Rate vs. K Value', fontsize=14)
    plt.xlabel('K', fontsize=12)
    plt.ylabel('Error Rate', fontsize=12)
    plt.grid(True, alpha=0.3)
    plt.show()

    # Find best K values
    max_train_score = max(train_scores)
    train_scores_ind = [i for i, v in enumerate(train_scores) if v == max_train_score]
    print(f'Max train score: {max_train_score*100:.2f}% at K = {[x+1 for x in train_scores_ind]}')

    max_test_score = max(test_scores)
    test_scores_ind = [i for i, v in enumerate(test_scores) if v == max_test_score]
    print(f'Max test score: {max_test_score*100:.2f}% at K = {[x+1 for x in test_scores_ind]}')

    return test_scores

    # Evaluate performance on unscaled data using the provided function
print("Performance on Unscaled Data:")
unchanged_test_scores = plot_KNN_error_rate(unchanged_data, diabetes_data['Outcome'])
```

## Standardize the Variables

Standardization (also called z-score normalization) is the process of putting different variables on the same scale. Standardization transforms your data such that the resulting distribution has a mean of 0 and a standard deviation of 1.

$$ Z = {X - \mu \over \sigma}$$

Where:
- Z is the standardized value
- X is the original value
- μ is the mean of the feature
- σ is the standard deviation of the feature

```python
from sklearn.preprocessing import StandardScaler

# Initialize the scaler
scaler = StandardScaler()

# Fit the scaler on training data features
scaler.fit(diabetes_data.drop('Outcome', axis=1))

# Transform the features
scaled_data = scaler.transform(diabetes_data.drop('Outcome', axis=1))
```

```python
# Create a dataframe with scaled features
df_feat = pd.DataFrame(scaled_data, columns=diabetes_data.columns[:-1])
df_feat.head()
```

```python
# Verify standardization: mean should be ~0, std should be ~1
print("Mean of scaled features:")
print(df_feat.mean())
print("\nStandard deviation of scaled features:")
print(df_feat.std())
```

```python
# Evaluate on scaled data
print("Performance on Standardized Data:")
scaled_test_scores = plot_KNN_error_rate(scaled_data, diabetes_data['Outcome'])
```

## Comparing Accuracy Before and After Standardization

```python
# Compare performance
plt.figure(figsize=(14, 8))
plt.plot(range(1, 40), unchanged_test_scores, marker='o', label='Unscaled Data', linewidth=2)
plt.plot(range(1, 40), scaled_test_scores, marker='s', label='Standardized Data', linewidth=2)
plt.title('KNN Accuracy: Unscaled vs Standardized Data', fontsize=14)
plt.xlabel('K Value', fontsize=12)
plt.ylabel('Test Accuracy', fontsize=12)
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.show()
```

# Conclusion

From the data analysis we carried out, it seems that there is some form of an association between BMI, number of pregnancies, glucose levels, and the test results for diabetes.

As for the classification tasks, the standardized data yields much better results than the unscaled data over most of the K-values considered, thus indicating the importance of standardizing data in Machine Learning problems. This improvement occurs because:

## Challenge 2: K-Fold Cross-Validation Analysis

**Objective:** Implement K-Fold cross-validation to get more robust performance estimates.

**Tasks:**
1. Import `cross_val_score` from sklearn.model_selection
2. Implement 5-fold cross-validation for KNN with K values from 1 to 40
3. Compare cross-validation scores for:
   - Unscaled data
   - Standardized data
   - MinMax-scaled data (from Challenge 1)
4. Plot the mean cross-validation accuracy with standard deviation error bars
5. Determine the optimal K value for each scaling approach

**Bonus:** Calculate and display the standard deviation of cross-validation scores to understand model stability.

```python
from sklearn.model_selection import cross_val_score

# 1. Initialize arrays to store cross-validation results
k_range = range(1, 40)
cv_unscaled_means, cv_unscaled_stds = [], []
cv_scaled_means, cv_scaled_stds = [], []
cv_minmax_means, cv_minmax_stds = [], []

# 2. Loop through K values and calculate 5-fold CV metrics
for k in k_range:
    knn = KNeighborsClassifier(n_neighbors=k)

    # Unscaled Data
    scores_unscaled = cross_val_score(knn, unchanged_data, y, cv=5)
    cv_unscaled_means.append(scores_unscaled.mean())
    cv_unscaled_stds.append(scores_unscaled.std())

    # Standardized Data
    scores_scaled = cross_val_score(knn, scaled_data, y, cv=5)
    cv_scaled_means.append(scores_scaled.mean())
    cv_scaled_stds.append(scores_scaled.std())

    # MinMax-Scaled Data
    scores_minmax = cross_val_score(knn, minmax_data, y, cv=5)
    cv_minmax_means.append(scores_minmax.mean())
    cv_minmax_stds.append(scores_minmax.std())

# 3. Plot the mean cross-validation accuracy with error bars (Standard Deviation)
plt.figure(figsize=(15, 8))

plt.errorbar(k_range, cv_unscaled_means, yerr=cv_unscaled_stds, fmt='-o', label='Unscaled Data', alpha=0.7, capsize=3)
plt.errorbar(k_range, cv_scaled_means, yerr=cv_scaled_stds, fmt='-s', label='Standardized Data (StandardScaler)', alpha=0.7, capsize=3)
plt.errorbar(k_range, cv_minmax_means, yerr=cv_minmax_stds, fmt='-^', label='MinMax-Scaled Data (MinMaxScaler)', alpha=0.7, capsize=3)

plt.title('5-Fold Cross-Validation Accuracy vs. K Value', fontsize=14)
plt.xlabel('K Value', fontsize=12)
plt.ylabel('Mean CV Accuracy', fontsize=12)
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.show()

# 4. Display optimal K and stability metrics
print("--- Optimal Performance Metrics (5-Fold CV) ---")
best_unscaled_idx = np.argmax(cv_unscaled_means)
print(f"Unscaled Data       -> Max Mean Accuracy: {cv_unscaled_means[best_unscaled_idx]*100:.2f}% at K = {best_unscaled_idx+1} (Std: {cv_unscaled_stds[best_unscaled_idx]*100:.2f}%)")

best_scaled_idx = np.argmax(cv_scaled_means)
print(f"Standardized Data   -> Max Mean Accuracy: {cv_scaled_means[best_scaled_idx]*100:.2f}% at K = {best_scaled_idx+1} (Std: {cv_scaled_stds[best_scaled_idx]*100:.2f}%)")

best_minmax_idx = np.argmax(cv_minmax_means)
print(f"MinMax-Scaled Data  -> Max Mean Accuracy: {cv_minmax_means[best_minmax_idx]*100:.2f}% at K = {best_minmax_idx+1} (Std: {cv_minmax_stds[best_minmax_idx]*100:.2f}%)")
```

```python
# Print stability metrics (Standard Deviation of CV Scores)
print("--- Model Stability Analysis (Standard Deviation) ---")
print(f"Unscaled Data       -> Mean Std Dev across all K: {np.mean(cv_unscaled_stds)*100:.2f}%")
print(f"Standardized Data   -> Mean Std Dev across all K: {np.mean(cv_scaled_stds)*100:.2f}%")
print(f"MinMax-Scaled Data  -> Mean Std Dev across all K: {np.mean(cv_minmax_stds)*100:.2f}%")

# Find the absolute most stable K value (lowest variance/std dev) for Standardized Data
most_stable_scaled_k = np.argmin(cv_scaled_stds) + 1
print(f"\nMost stable configuration for Standardized Data is at K = {most_stable_scaled_k} "
      f"with a standard deviation of {cv_scaled_stds[most_stable_scaled_k-1]*100:.2f}%")
```

```python
# 1. Define the top 4 correlated features from Task 1
top_features = ['Glucose', 'BMI', 'Age', 'Pregnancies']

# 2. Extract only these features from the standardized DataFrame
X_train_top = pd.DataFrame(scaled_data, columns=diabetes_data.columns[:-1])[top_features]

# 3. Perform a 5-fold cross-validation across K values using only top features
cv_top_means = []
for k in range(1, 40):
    knn_top = KNeighborsClassifier(n_neighbors=k)
    scores_top = cross_val_score(knn_top, X_train_top, y, cv=5)
    cv_top_means.append(scores_top.mean())

# 4. Plot the subset model performance against the full standardized model
plt.figure(figsize=(14, 7))
plt.plot(range(1, 40), cv_scaled_means, marker='s', label='All 8 Features (Standardized)', color='orange', linewidth=2)
plt.plot(range(1, 40), cv_top_means, marker='o', label='Top 4 Features Only (Standardized)', color='purple', linewidth=2)

plt.title('Feature Subset Analysis: 8 Features vs. Top 4 Features', fontsize=14)
plt.xlabel('K Value', fontsize=12)
plt.ylabel('Mean CV Accuracy', fontsize=12)
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.show()

# 5. Output comparison results
best_top_idx = np.argmax(cv_top_means)
print(f"All 8 Features Model -> Max Mean CV Accuracy: {max(cv_scaled_means)*100:.2f}%")
print(f"Top 4 Features Model -> Max Mean CV Accuracy: {cv_top_means[best_top_idx]*100:.2f}% at K = {best_top_idx+1}")
```

```python
# 1. Define feature subsets based on Task 1 correlation rankings
# Top features ranked: Glucose, BMI, Age, Pregnancies, SkinThickness, Insulin...
features_ranked = ['Glucose', 'BMI', 'Age', 'Pregnancies', 'SkinThickness', 'Insulin']

subsets = {
    'Top 3 Features': features_ranked[:3],   # ['Glucose', 'BMI', 'Age']
    'Top 4 Features': features_ranked[:4],   # ['Glucose', 'BMI', 'Age', 'Pregnancies']
    'Top 5 Features': features_ranked[:5],   # Includes SkinThickness
    'Top 6 Features': features_ranked[:6]    # Includes Insulin
}

# Convert scaled data back to a DataFrame for easy slicing
df_scaled_features = pd.DataFrame(scaled_data, columns=diabetes_data.columns[:-1])

plt.figure(figsize=(15, 8))

# 2. Loop through each subset, run 5-fold CV, and plot the results
for label, feature_list in subsets.items():
    cv_means = []
    X_subset = df_scaled_features[feature_list]

    for k in range(1, 40):
        knn_model = KNeighborsClassifier(n_neighbors=k)
        scores = cross_val_score(knn_model, X_subset, y, cv=5)
        cv_means.append(scores.mean())

    # Plot this specific subset's performance curve
    plt.plot(range(1, 40), cv_means, marker='o', label=f'{label} ({len(feature_list)} features)', linewidth=2)

    # Print out peak performance details for accountability
    best_k_idx = np.argmax(cv_means)
    print(f"{label} -> Max Mean CV Accuracy: {cv_means[best_k_idx]*100:.2f}% at K = {best_k_idx+1}")

# 3. Add the baseline Full 8-Feature model to the plot for a complete picture
plt.plot(range(1, 40), cv_scaled_means, marker='s', label='All 8 Features (Baseline)', color='black', linestyle='--', linewidth=2)
print(f"All 8 Features   -> Max Mean CV Accuracy: {max(cv_scaled_means)*100:.2f}%")

plt.title('The Impact of Feature Dimensions on KNN Performance', fontsize=14)
plt.xlabel('K Value', fontsize=12)
plt.ylabel('Mean CV Accuracy', fontsize=12)
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.show()
```
